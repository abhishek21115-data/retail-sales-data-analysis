```python
import pandas as pd

df = pd.read_excel("Online Retail.xlsx")
df.head()
```

```python
# Check how much data we have before cleaning
print("Before cleaning:", df.shape)

# Remove rows where CustomerID is missing
df = df.dropna(subset=["CustomerID"])

# Remove cancelled orders (InvoiceNo starting with "C")
df = df[~df["InvoiceNo"].astype(str).str.startswith("C")]

# Remove rows with negative or zero quantity
df = df[df["Quantity"] > 0]

# Check how much data remains after cleaning
print("After cleaning:", df.shape)
```

```python
# Create a new column: total price per row
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]

# Peek at the result
df[["InvoiceNo", "Quantity", "UnitPrice", "TotalPrice"]].head()
```

```python
# Save the cleaned data as a CSV file
df.to_csv("cleaned_online_retail.csv", index=False)

# Download it to your computer
from google.colab import files
files.download("cleaned_online_retail.csv")
```
